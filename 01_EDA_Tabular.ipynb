Missing Values

```python
missing = df.isnull().sum()
missing_percent = (df.isnull().sum() / len(df)) * 100

missing_df = pd.DataFrame({
    'Missing Values': missing,
    'Percentage (%)': missing_percent
})

print('Missing Values Report:')
print(missing_df[missing_df['Missing Values'] > 0])

if missing.sum() == 0:
    print('No missing values! Dataset is clean.')
else:
    print(f' Total missing values: {missing.sum()}')  
```

Missing Values Heatmap

```python
plt.figure(figsize=(12, 5))
sns.heatmap(df.isnull(), 
            yticklabels=False, 
            cbar=True, 
            cmap='viridis')
plt.title('Missing Values Heatmap', fontsize=14, fontweight='bold')

# Yeh line add karo — column names horizontal ho jayenge
plt.xticks(rotation=45, ha='right', fontsize=8)

plt.tight_layout()
plt.savefig('missing_values_heatmap.png', dpi=150, bbox_inches='tight')
plt.show()
print('Plot saved!')
```

Target Column and Charts

```python
target_col = 'Diagnosis'

if target_col not in df.columns:
    target_col = df.columns[-1]
    print(f'Using last column: {target_col}')

print(f'Target Column: {target_col}')
print(df[target_col].value_counts())
print(df[target_col].value_counts(normalize=True) * 100)

df[target_col] = df[target_col].map({0: 'Non-Demented', 1: 'Demented'})

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes = np.array(axes).flatten()

value_counts = df[target_col].value_counts()
colors = sns.color_palette('husl', len(value_counts))

# Bar chart
axes[0].bar(value_counts.index.astype(str),
            value_counts.values,
            color=colors,
            edgecolor='black',
            linewidth=0.5)
axes[0].set_title('Class Distribution - Bar Chart', 
                  fontsize=13, fontweight='bold')
axes[0].set_xlabel('Diagnosis', fontsize=11)
axes[0].set_ylabel('Count', fontsize=11)
for i, v in enumerate(value_counts.values):
    axes[0].text(i, v + 1, str(v), ha='center', fontweight='bold')

# Pie chart
axes[1].pie(value_counts.values,
            labels=value_counts.index.astype(str),
            autopct='%1.1f%%',
            colors=colors,
            startangle=90)
axes[1].set_title('Class Distribution - Pie Chart', 
                  fontsize=13, fontweight='bold')

plt.suptitle('Alzheimer Disease - Target Class Distribution',
             fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('class_distribution.png', dpi=150, bbox_inches='tight')
plt.show()
print('Plot saved!')
```

Feature Distribution Histogram

```python
numerical_cols = df.select_dtypes(include=[np.number]).columns.tolist()
categorical_cols = df.select_dtypes(include=['object']).columns.tolist()

# Target column remove karo agar numerical mein hai
if target_col in numerical_cols:
    numerical_cols.remove(target_col)

print(f'Numerical Columns ({len(numerical_cols)}): {numerical_cols}')
print(f'Categorical Columns ({len(categorical_cols)}): {categorical_cols}')

# Histograms
n_cols = 3
n_rows = (len(numerical_cols) + n_cols - 1) // n_cols

fig, axes = plt.subplots(n_rows, n_cols, figsize=(15, n_rows * 4))
axes = axes.flatten()

for i, col in enumerate(numerical_cols):
    axes[i].hist(df[col].dropna(), bins=30, 
                 color='steelblue', edgecolor='black', alpha=0.7)
    axes[i].set_title(f'{col}', fontsize=11, fontweight='bold')
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Frequency')

# Extra axes hide karo
for j in range(i + 1, len(axes)):
    axes[j].set_visible(False)

plt.suptitle('Numerical Features Distribution', 
             fontsize=14, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('numerical_distributions.png', dpi=150, bbox_inches='tight')
plt.show()
print('Plot saved!')
```

Boxplots

```python
cols_to_plot = [col for col in numerical_cols 
                if col not in ['PatientID', 'DoctorInCharge']][:6]
fig, axes = plt.subplots(2, 3, figsize=(16, 10))
axes = axes.flatten()
for i, col in enumerate(cols_to_plot):
    df.boxplot(column=col, by=target_col, ax=axes[i])
    axes[i].set_title(f'{col} by Diagnosis', 
                      fontsize=11, fontweight='bold')
    axes[i].set_xlabel('Diagnosis')
    axes[i].set_ylabel(col)

plt.suptitle('Feature Distribution by Diagnosis - Boxplots',
             fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('boxplots.png', dpi=150, bbox_inches='tight')
plt.show()
print('Boxplot saved!')
```

Correlation Heatmap

```python
plt.figure(figsize=(14, 10))
corr_matrix = df[numerical_cols].corr()

mask = np.triu(np.ones_like(corr_matrix, dtype=bool))

sns.heatmap(corr_matrix,
            mask=mask,
            annot=True,
            fmt='.2f',
            cmap='coolwarm',
            center=0,
            square=True,
            linewidths=0.5,
            cbar_kws={'shrink': 0.8})

plt.title('Feature Correlation Heatmap', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.savefig('correlation_heatmap.png', dpi=150, bbox_inches='tight')
plt.show()
print('Heatmap saved!')
```

Duplicates and Summary

```python
duplicates = df.duplicated().sum()
print(f'Duplicate Rows: {duplicates}')

if duplicates > 0:
    df = df.drop_duplicates()
    print(f'Duplicates removed! New shape: {df.shape}')
else:
    print('No duplicates found!')

print('=' * 50)
print('TABULAR EDA COMPLETE!')
print(f'Total Patients  : {df.shape[0]}')
print(f'Total Features  : {df.shape[1]}')
print(f'Missing Values  : {df.isnull().sum().sum()}')
print(f'Duplicate Rows  : {df.duplicated().sum()}')
print('=' * 50)
```
