Libraries

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
import warnings
warnings.filterwarnings('ignore')

print('Libraries imported!')
```

Loading CSV File

```python
path = r'C:\Users\nbush\OneDrive - University of Huddersfield\Desktop\FYP\data_tabular\alzheimers_disease_data.csv'

df = pd.read_csv(path)

demented = df[df['Diagnosis'] == 1]
normal   = df[df['Diagnosis'] == 0]

cols = ['Age', 'BMI', 'SystolicBP', 'DiastolicBP',
        'CholesterolTotal', 'MMSE',
        'FunctionalAssessment', 'ADL',
        'SleepQuality', 'DietQuality',
        'AlcoholConsumption', 'PhysicalActivity',
        'MemoryComplaints', 'BehavioralProblems',
        'Confusion', 'Disorientation',
        'PersonalityChanges',
        'DifficultyCompletingTasks',
        'Forgetfulness', 'Depression',
        'Smoking', 'FamilyHistoryAlzheimers',
        'CardiovascularDisease', 'Diabetes',
        'HeadInjury', 'Hypertension']

print('DEMENTED average values:')
print(demented[cols].mean().round(2))
print('\nNORMAL average values:')
print(normal[cols].mean().round(2))
```

Dropping Unnecessary Columns

```python
# PatientID aur DoctorInCharge useless hain
cols_to_drop = ['PatientID', 'DoctorInCharge']
df = df.drop(columns=cols_to_drop)

print(f'Columns dropped: {cols_to_drop}')
print(f'New shape: {df.shape}')
```

Categorical Encoding

```python
# Categorical columns check karo
cat_cols = df.select_dtypes(include=['object']).columns.tolist()
print(f'Categorical columns: {cat_cols}')

# Label Encoding
le = LabelEncoder()
for col in cat_cols:
    df[col] = le.fit_transform(df[col])
    print(f'{col} encoded!')

print(f'\nData after encoding:')
df.head()
```

Separating Feature and Target

```python
# Target column
target_col = 'Diagnosis'

# X = features, y = target
X = df.drop(columns=[target_col])
y = df[target_col]

print(f'Features (X) shape: {X.shape}')
print(f'Target (y) shape: {y.shape}')
print(f'\nTarget distribution:')
print(y.value_counts())
```

Train\Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,      # 20% test
    random_state=42,    # Reproducible results
    stratify=y          # Class balance maintain karo
)

print(f'Split complete!')
print(f'Training set  : {X_train.shape}')
print(f'Testing set   : {X_test.shape}')
print(f'\nTraining target distribution:')
print(y_train.value_counts())
print(f'\nTesting target distribution:')
print(y_test.value_counts())
```

Feature Scaling

```python
scaler = StandardScaler()

# Sirf training data par fit karo
X_train_scaled = scaler.fit_transform(X_train)

# Test data par sirf transform karo
X_test_scaled = scaler.transform(X_test)

# DataFrame mein convert karo
X_train_scaled = pd.DataFrame(
    X_train_scaled,
    columns=X_train.columns
)
X_test_scaled = pd.DataFrame(
    X_test_scaled,
    columns=X_test.columns
)

print(f'Scaling complete!')
print(f'\nBefore scaling:')
print(f'Age mean: {X_train["Age"].mean():.2f}')
print(f'\nAfter scaling:')
print(f'Age mean: {X_train_scaled["Age"].mean():.4f}')
```

Saving Data

```python
import os

# Preprocessed folder banao
save_path = r'C:\Users\nbush\OneDrive - University of Huddersfield\Desktop\FYP\preprocessed'
os.makedirs(save_path, exist_ok=True)

# Save karo
X_train_scaled.to_csv(f'{save_path}\\X_train_tabular.csv', index=False)
X_test_scaled.to_csv(f'{save_path}\\X_test_tabular.csv', index=False)
y_train.to_csv(f'{save_path}\\y_train_tabular.csv', index=False)
y_test.to_csv(f'{save_path}\\y_test_tabular.csv', index=False)

print(f'All files saved to preprocessed folder!')
print(f'\nFiles saved:')
print(f'  → X_train_tabular.csv')
print(f'  → X_test_tabular.csv')
print(f'  → y_train_tabular.csv')
print(f'  → y_test_tabular.csv')
```
