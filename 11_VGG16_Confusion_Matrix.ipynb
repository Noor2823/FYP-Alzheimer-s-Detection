Libraries

```python
import os
os.environ['TF_ENABLE_ONEDNN_OPTS'] = '0'
os.environ['TF_CPP_MIN_LOG_LEVEL'] = '3'

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import tensorflow as tf
from sklearn.metrics import confusion_matrix

print(' Libraries imported!')
```

Loading Data and Model

```python
path = r'C:\Users\nbush\OneDrive - University of Huddersfield\Desktop\FYP\preprocessed'

X_test = np.load(f'{path}\\X_test_images.npy')
y_test = np.load(f'{path}\\y_test_images.npy')

model = tf.keras.models.load_model(
    f'{path}\\best_vgg16_model.keras'
)

classes = ['MildDemented', 'ModerateDemented',
           'NonDemented', 'VeryMildDemented']

print(' Loaded!')
```

Confusion Matrix

```python
y_pred_proba = model.predict(X_test)
y_pred = np.argmax(y_pred_proba, axis=1)

cm = confusion_matrix(y_test, y_pred)

plt.figure(figsize=(8, 6))
sns.heatmap(cm,
            annot=True,
            fmt='d',
            cmap='Oranges',
            xticklabels=classes,
            yticklabels=classes,
            linewidths=0.5)

plt.title('VGG16 — Confusion Matrix',
          fontsize=13, fontweight='bold')
plt.xlabel('Predicted', fontsize=11)
plt.ylabel('Actual', fontsize=11)
plt.xticks(rotation=15)
plt.tight_layout()
plt.savefig(
    r'C:\Users\nbush\OneDrive - University of Huddersfield\Desktop\FYP\notebooks\vgg16_confusion_matrix.png',
    dpi=150, bbox_inches='tight')
plt.show()
print(' Saved!')
```
